# Analyst 101 — a Jupyter notebook running on Databricks

This is an ordinary `.ipynb` file — the same format you'd open in JupyterLab — **imported straight into the Databricks workspace** and run on serverless compute. Nothing was rewritten: your Jupyter skills transfer as-is.

Below we point at the raw landing file, explore it, and see why *documented* data is what makes Genie and the AI/BI Assistant accurate.

*Synthetic data — no PHI.*

```python
# --- point at your uploaded landing file (edit these to your schema) ---
CATALOG = "rch_analyst101"
SCHEMA  = "analyst101_shared"      # the schema that holds the `landing` volume + raw CSV
PROFILE = "pediatric"              # which facilities_raw.<profile>.csv you uploaded

csv_path = f"/Volumes/{CATALOG}/{SCHEMA}/landing/facilities_raw.{PROFILE}.csv"
print("Reading landing file:", csv_path)
```

## Explore the raw landing file with `read_files()`

Scan the CSV directly — no ingestion step needed. Notice it's messy on purpose: padded text, `State` in mixed forms, inconsistent `Type` casing, blank `Region`s, and a duplicate row. This is the **bronze** (as-ingested) layer.

```python
raw = spark.sql(f"""
    SELECT *
    FROM read_files('{csv_path}', format => 'csv', header => 'true')
""")
display(raw)
```

## A quick look — the way you would in Jupyter

Standard exploration: row count and the distinct `State` values (spot the inconsistent casing/spacing).

```python
print("Row count:", raw.count())
display(raw.select("State").distinct().orderBy("State"))
```

## Why this matters for Genie — discoverability

Genie and the AI/BI Assistant answer well when the data is **documented**: table + column comments, primary/foreign keys, and clear names. A raw file like this has none of that yet.

In the workshop you turn this into a governed `dim_facility` and add `COMMENT ON` / `PRIMARY KEY` — and that documentation is exactly what a Genie Agent reads to generate correct SQL. Same notebook you'd run in Jupyter, now on a governed platform.
